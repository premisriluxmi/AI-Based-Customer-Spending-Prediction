# AI-Based Customer Spending Prediction

## Day 1: Dataset Understanding and Preparation

### Objective
To create, load, and understand a customer spending dataset
that will be used to predict customer spending categories:
Low, Medium, and High.

```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
np.random.seed(42)

n = 200

data = {
    "Customer_ID": range(1, n + 1),
    "Age": np.random.randint(18, 61, n),
    "Annual_Income": np.random.randint(150000, 1500001, n),
    "Purchase_Frequency": np.random.randint(1, 16, n),
    "Avg_Order_Value": np.random.randint(500, 5001, n),
    "Membership_Years": np.random.randint(1, 11, n),
    "Website_Visits": np.random.randint(1, 31, n)
}

df = pd.DataFrame(data)

# Calculate estimated monthly spending
df["Monthly_Spending"] = (
    df["Purchase_Frequency"] * df["Avg_Order_Value"]
)

# Create spending categories
def spending_category(amount):
    if amount < 10000:
        return "Low"
    elif amount <= 25000:
        return "Medium"
    else:
        return "High"

df["Spending_Category"] = df["Monthly_Spending"].apply(
    spending_category
)

print("Dataset created successfully!")
```

```python
df.to_csv("customer_spending_dataset.csv", index=False)

print("Dataset saved successfully!")
```
